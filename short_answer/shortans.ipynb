```python
import zipfile
import pandas as pd

with zipfile.ZipFile("train.tsv.zip", "r") as zip_ref:
    zip_ref.extractall()

df = pd.read_csv(
    "train.tsv",
    sep="\t",
    encoding="latin1"
)

print(df.shape)
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
X = df['EssayText']
y = df['Score1']
```

```python
vectorizer = TfidfVectorizer(
    max_features=10000,
    stop_words='english'
)

X_vec = vectorizer.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_vec,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=2000
)

model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

preds = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, preds))
```

```python
import joblib

joblib.dump(model, "assignment_evaluator.pkl")
joblib.dump(vectorizer, "vectorizer.pkl")

print("Saved Successfully")
```

```python
sample_answer = """
Plants need sunlight because sunlight helps them make food through photosynthesis.
Without sunlight the plant cannot produce enough energy to survive.
"""

sample_vec = vectorizer.transform([sample_answer])

prediction = model.predict(sample_vec)

print("Predicted Score:", prediction[0])
```

```python
df["combined_text"] = (
    "EssaySet_" +
    df["EssaySet"].astype(str) +
    " " +
    df["EssayText"]
)
```

```python
X = df["combined_text"]
y = df["Score1"]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer2 = TfidfVectorizer(
    max_features=15000,
    stop_words="english"
)

X_vec = vectorizer2.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_vec,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.linear_model import LogisticRegression

model2 = LogisticRegression(
    max_iter=3000
)

model2.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

preds2 = model2.predict(X_test)

print("Accuracy:", accuracy_score(y_test, preds2))
```

```python
import joblib

joblib.dump(model2, "assignment_evaluator.pkl")
joblib.dump(vectorizer2, "vectorizer.pkl")
```
